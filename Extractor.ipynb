```python
!pip install mediapipe
from google.colab import drive
import os
import cv2
import numpy as np
import mediapipe as mp
import pandas as pd
from google.colab.patches import cv2_imshow
```

```python
# Mount Google Drive
drive.mount('/content/gdrive')
```

```python
# Set up MediaPipe Pose model
mp_pose = mp.solutions.pose
pose = mp_pose.Pose(
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)
```

```python
# Set up MediaPipe drawing utilities
mp_drawing = mp.solutions.drawing_utils
```

```python
# Specify the folder containing the images in Google Drive
image_folder = '/content/gdrive/MyDrive/hkey'
# Initialize an empty list to store the keypoints
all_keypoints = []
```

```python
# Loop through the images in the folder
for filename in os.listdir(image_folder):
    if filename.endswith('.jpg') or filename.endswith('.png'):
        # Load the image
        image_path = os.path.join(image_folder, filename)
        image = cv2.imread(image_path)

        # Process the image and extract keypoints
        results = pose.process(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))

        # Draw the keypoints on the image
        image_with_keypoints = image.copy()
        if results.pose_landmarks:
            mp_drawing.draw_landmarks(
                image_with_keypoints,
                results.pose_landmarks,
                mp_pose.POSE_CONNECTIONS,
                landmark_drawing_spec=mp_drawing.DrawingSpec(color=(255, 255, 255), thickness=2, circle_radius=2),
                connection_drawing_spec=mp_drawing.DrawingSpec(color=(0, 255, 0), thickness=2)
            )

        # Display the image with keypoints
        cv2_imshow(image_with_keypoints)

        # Store the keypoints in the list
        if results.pose_world_landmarks:
            keypoints = results.pose_world_landmarks.landmark
            keypoints_np = np.array([[keypoint.x, keypoint.y, keypoint.z] for keypoint in keypoints])
            all_keypoints.append(keypoints_np)
```

```python

# Save the keypoints to a CSV file in Google Drive
keypoints_df = pd.DataFrame(np.concatenate(all_keypoints, axis=0))
keypoints_df.to_csv('/content/gdrive/MyDrive/rrrrrr.csv', index=False, header=False)
print("TRUE")
```
